```python
import os
import shutil

# Define possible extensions
extensions = ['.jpg', '.JPG', '.png']

# Iterate over the DataFrame rows
for index, row in df_move.iterrows():
    img_folder = row['img_path'] 
    file_found = False

    for ext in extensions:
        modified_img_name = row['img_name'] + ext  # Append extension
        source_path = os.path.join(basepath, img_folder, modified_img_name)

        # Construct the destination file path
        dest_file_path = os.path.join(destination_path, modified_img_name)

        # Check if the image has already been copied
        if modified_img_name not in copied_images:
            # Try copying the file with the current extension
            if os.path.exists(source_path):
                shutil.copy2(source_path, dest_file_path)
                copied_images.add(modified_img_name)
                file_found = True
                break  # Stop trying other extensions
            else:
                continue  # Try the next extension
        else:
            print(f"Already copied: {modified_img_name}")
            file_found = True
            break  # File already copied, no need to try other extensions

    if not file_found:
        print(f"File not found for any extensions: {row['img_name']}")
```

```python
# 好像是remove from 某一个folder，if file exists in a list?
# 我不记得了

import os
import shutil

def get_modified_jpg_filenames_in_directory(directory, remove_first_n=0):
    jpg_files = []
    for filename in os.listdir(directory):
        if filename.lower().endswith(('.jpg', '.jpeg')):
            modified_filename = filename[remove_first_n:]  # Remove first n characters
            jpg_files.append(modified_filename)
    return jpg_files

def compare_and_copy_images(LI_filenames, Yolo_filenames, source_dir_LI, source_dir_Yolo, destination_dir):
    # Create a set of filenames for efficient comparison
    LI_set = set(LI_filenames)
    Yolo_set = set(Yolo_filenames)

    # Find images that only exist in Yolo_filenames
    unique_to_yolo = Yolo_set - LI_set

    # Copy images unique to Yolo_filenames to destination directory
    for filename in unique_to_yolo:
        source_path = os.path.join(source_dir_Yolo, filename)
        destination_path = os.path.join(destination_dir, filename)
        shutil.copyfile(source_path, destination_path)

    # Find images that only exist in LI_filenames
    unique_to_li = LI_set - Yolo_set

    # Save images unique to LI_filenames to a list
    unique_to_li_list = list(unique_to_li)

    # Print images unique to LI_filenames
    print("Images unique to LI_filenames:")
    for filename in unique_to_li_list:
        print(filename)
    
    return unique_to_li_list

# Usage:
source_dir_LI = '/Users/wenlanzhang/Downloads/images/'
source_dir_Yolo = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/GoogleStreetView/Yolo_Train_Waste_Major/'
destination_dir = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/GoogleStreetView/tobelabel/'

LI_filenames = get_modified_jpg_filenames_in_directory(source_dir_LI, remove_first_n=9)
Yolo_filenames = get_modified_jpg_filenames_in_directory(source_dir_Yolo)

unique_to_li_list = compare_and_copy_images(LI_filenames, Yolo_filenames, source_dir_LI, source_dir_Yolo, destination_dir)
```
