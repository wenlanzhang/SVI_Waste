```python
# Yolo8 Only
# Reshape the DataFrame from wide to long format
df_long = pd.melt(df_GSVI_sorted, id_vars=['index', 'panoid', 'lat', 'lon', 'year', 'month', 'img_path'],
                   var_name='angle', value_name='value')

# Extract the angle values from the column names
df_long['angle'] = df_long['angle'].str.replace('ang_', '').astype(int)

# Print the resulting long-format DataFrame
df_long
```

```python
# Yolov8 and LLM 

# Step 1: Melt for Yo columns
yo_long = pd.melt(df_GSVI_sorted, id_vars=['index', 'panoid', 'lat', 'lon', 'year', 'month', 'img_path'],
                  value_vars=['Yo_0', 'Yo_90', 'Yo_180', 'Yo_270'],
                  var_name='yo_angle', value_name='Yo')

# Extract angle for Yo
yo_long['ang'] = yo_long['yo_angle'].str.extract('(\d+)$').astype(int)
yo_long.drop('yo_angle', axis=1, inplace=True)

# Step 2: Melt for LLM columns
llm_long = pd.melt(df_GSVI_sorted, id_vars=['index', 'panoid', 'lat', 'lon', 'year', 'month', 'img_path'],
                   value_vars=['LLM_0', 'LLM_90', 'LLM_180', 'LLM_270'],
                   var_name='llm_angle', value_name='LLM')

# Extract angle for LLM
llm_long['ang'] = llm_long['llm_angle'].str.extract('(\d+)$').astype(int)
llm_long.drop('llm_angle', axis=1, inplace=True)

# Step 3: Merge the Yo and LLM long DataFrames
df_long = pd.merge(yo_long, llm_long, on=['index', 'panoid', 'lat', 'lon', 'year', 'month', 'img_path', 'ang'], how='inner')

# The resulting df_long DataFrame now has separate Yo and LLM columns, with a common 'ang' column for the angle
df_long
```

```python
df_long.drop(columns=['index'], inplace=True)

# df_long = df_long.drop_duplicates() # duplicates should have been considered at the beginning.
df_long
```

```python
# Set your base path for the source images
basepath = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/GoogleStreetView/Maoran/img/'  # Replace with your base path

# Set your destination path
destination_path = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/GoogleStreetView/PreScreen/'  # Replace with your destination path

# Set to keep track of copied images
copied_images = set()
```

```python
# Iterate over the DataFrame rows
for index, row in df_long_se.iterrows():
    # Construct the source file path
    img_folder = row['img_path']  # This should be something like 'Q/g/'
    modified_img_name = row['img_name'] + '.jpg'  # Assuming the file extension is .jpg
    source_path = os.path.join(basepath, img_folder, modified_img_name)

    # Construct the destination file path
    dest_file_path = os.path.join(destination_path, modified_img_name)

    # Check if the image has already been copied
    if modified_img_name not in copied_images:
        # Copy the file
        if os.path.exists(source_path):
            shutil.copy2(source_path, dest_file_path)
            copied_images.add(modified_img_name)
        else:
            print(f"File not found: {source_path}")
    else:
        print(f"Already copied: {modified_img_name}")
```
