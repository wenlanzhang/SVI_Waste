```python
def extract_gps_coordinates(image_path):
    """
    Extract and convert GPS coordinates from an image's EXIF metadata.
    """
    def dms_to_decimal(dms, ref):
        """
        Convert GPS DMS (degrees, minutes, seconds) to decimal degrees.
        """
        degrees, minutes, seconds = dms
        decimal = degrees + (minutes / 60.0) + (seconds / 3600.0)
        if ref in ['S', 'W']:  # South and West are negative
            decimal = -decimal
        return decimal

    # Open the image and extract EXIF data
    image = Image.open(image_path)
    exif_data = image._getexif()
    if not exif_data:
        return None, None

    # Extract GPSInfo
    gps_info_raw = exif_data.get(34853)  # 34853 is the GPSInfo tag ID
    if not gps_info_raw:
        return None, None

    # Map GPSInfo keys to human-readable tags
    gps_info = {GPSTAGS.get(k, k): v for k, v in gps_info_raw.items()}

    # Extract and convert GPS coordinates
    latitude = gps_info.get("GPSLatitude")
    latitude_ref = gps_info.get("GPSLatitudeRef")
    longitude = gps_info.get("GPSLongitude")
    longitude_ref = gps_info.get("GPSLongitudeRef")

    if latitude and latitude_ref and longitude and longitude_ref:
        lat_decimal = dms_to_decimal(latitude, latitude_ref)
        lon_decimal = dms_to_decimal(longitude, longitude_ref)
        return lat_decimal, lon_decimal
    else:
        return None, None

def extract_datetime(image_path):
    """
    Extract the datetime information from an image's EXIF metadata.
    """
    image = Image.open(image_path)
    exif_data = image._getexif()
    if not exif_data:
        return None

    # Extract DateTimeOriginal
    date_time = exif_data.get(36867)  # 36867 is the DateTimeOriginal tag ID
    return date_time

def process_images(image_dir):
    """
    Process all images in a directory and save metadata to a DataFrame.
    """
    data = []
    for file_name in os.listdir(image_dir):
        if file_name.lower().endswith(('.jpg', '.jpeg', '.png')):
            image_path = os.path.join(image_dir, file_name)

            # Extract GPS coordinates
            lat, lon = extract_gps_coordinates(image_path)

            # Extract datetime
            date_time = extract_datetime(image_path)
            if date_time:
                year, month, day = date_time.split(" ")[0].split(":")
                hour = date_time.split(" ")[1].split(":")[0]
            else:
                year, month, day, hour = None, None, None, None

            # Remove file extension from img_name
            img_name = os.path.splitext(file_name)[0]

            # Append data
            data.append({
                "img_name": img_name,  # Without the file extension
                "year": year,
                "month": month,
                "day": day,
                "hour": hour,
                "lat": lat,
                "lon": lon
            })

    # Create DataFrame
    df = pd.DataFrame(data)
    return df
```

```python
# Read the CSV file
file_path = base_path + "/pano_nairobi.csv"
df = pd.read_csv(file_path)

# Drop specific columns
columns_to_drop = ["Unnamed: 0", "id", "marker"]  # Replace with column names you want to delete
df = df.drop(columns=columns_to_drop)

columns_to_convert = ["year", "month"]
df[columns_to_convert] = df[columns_to_convert].astype(int)

# Display the first few rows of the DataFrame
df
```

```python
# Expand each row 4 times and add '_0', '_90', '_180', '_270' to panoid
df_expanded = df.loc[df.index.repeat(4)].reset_index(drop=True)  # Repeat each row 4 times
df_expanded["img_name"] = df_expanded["panoid"] + ["_0", "_90", "_180", "_270"] * (len(df) // 1)

# Create the img_dir based on the first and second letters of img_name
df_expanded["img_dir"] = df_expanded["img_name"].apply(
    lambda name: f"Google/{name[0]}/{name[1]}/")

# Check whether the image exists
def check_image_existence(row):
    img_full_path_jpg = os.path.join(base_path, row["img_dir"], row["img_name"] + ".jpg")
    img_full_path_JPG = os.path.join(base_path, row["img_dir"], row["img_name"] + ".JPG")
    
    # Check for both .jpg and .JPG
    if os.path.exists(img_full_path_jpg):
        return img_full_path_jpg, True
    elif os.path.exists(img_full_path_JPG):
        return img_full_path_JPG, True
    else:
        return None, False

# Apply the check function to populate img_full_path and exist columns
df_expanded[["img_full_path", "exist"]] = df_expanded.apply(
    lambda row: pd.Series(check_image_existence(row)), axis=1)

df_expanded
```

```python
df_expanded = df_expanded.drop(columns=["img_full_path"])
df_expanded["day"] = np.nan  # Set 'day' to NaN
df_expanded["hour"] = np.nan  # Set 'hour' to NaN

df_expanded.to_csv(base_path + 'GSVI_basic.csv', index=False)

df_expanded
```
