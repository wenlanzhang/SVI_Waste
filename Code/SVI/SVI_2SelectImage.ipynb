```python
import os
import shutil
import pandas as pd
```

```python
# Define the source name, change this to 'Google', 'Faith', or 'ZWL' as needed
source_name = 'Faith'  # Change this value as needed

# Define base path
base_path = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/Waste/img/'

# Define selected paths dynamically
yolo_selected_path = os.path.join(base_path, 'Yolo2_Selected', source_name)
cpm_selected_path = os.path.join(base_path, 'cpm2_Selected', source_name)
dif_path = os.path.join(base_path, 'yolo2_cpm_dif', source_name)

# Create directories dynamically
os.makedirs(yolo_selected_path, exist_ok=True)
os.makedirs(cpm_selected_path, exist_ok=True)
os.makedirs(dif_path, exist_ok=True)

# Load the CSV file dynamically
csv_file_path = os.path.join(base_path, f'cpm2_{source_name}.csv')
df = pd.read_csv(csv_file_path)
df
```

# Copy yolo and cpm output to folder 

```python
# Function to copy images based on conditions
def copy_images(row):
    img_name = row['img_name']
    img_dir = row['img_dir'] 
    
    # Construct the full image path without extension
    img_path_without_ext = os.path.join(base_path, img_dir, img_name)
    
    # Check for common image extensions (e.g., .jpg, .JPG, .png, etc.)
    extensions = ['.jpg', '.JPG', '.jpeg', '.png', '.PNG']  # Add more if needed
    img_path = None
    
    for ext in extensions:
        candidate_path = img_path_without_ext + ext
        if os.path.exists(candidate_path):
            img_path = candidate_path
            break
    
    # If no image is found, print a warning and skip
    if img_path is None:
        print(f"Image {img_name} not found in {os.path.join(base_path, img_dir)} with any supported extension.")
        return
    
    # Copy images where yolo_conf is not empty
    if row['yolo_conf'] != '[]':  # Check if yolo_conf is not empty
        shutil.copy(img_path, os.path.join(yolo_selected_path, os.path.basename(img_path)))
    
    # Copy images where prediction is 'yes'
    if row['prediction'] == 'yes':
        shutil.copy(img_path, os.path.join(cpm_selected_path, os.path.basename(img_path)))

# Apply the function to each row in the DataFrame
df.apply(copy_images, axis=1)

print("Copying completed!")
```

# Copy yolo and cpm diff output to folder 

```python
# Function to copy images where prediction == 'no'
def copy_images_prediction_no(row):
    img_name = row['img_name']
    img_dir = row['img_dir'] 
    
    # Construct the full image path without extension
    img_path_without_ext = os.path.join(base_path, img_dir, img_name)
    
    # Check for common image extensions (e.g., .jpg, .JPG, .png, etc.)
    extensions = ['.jpg', '.JPG', '.jpeg', '.png', '.PNG']  # Add more if needed
    img_path = None
    
    for ext in extensions:
        candidate_path = img_path_without_ext + ext
        if os.path.exists(candidate_path):
            img_path = candidate_path
            break
    
    # If no image is found, print a warning and skip
    if img_path is None:
        print(f"Image {img_name} not found in {os.path.join(base_path, img_dir)} with any supported extension.")
        return
    
    # Copy images where prediction == 'no' to path3
    if row['prediction'] == 'no':
        shutil.copy(img_path, os.path.join(dif_path, os.path.basename(img_path)))

# Apply the function to each row in the DataFrame
df.apply(copy_images_prediction_no, axis=1)

print("Copying images where prediction == 'no' completed!")
```

# Copy QWen output to folder

```python
import os
import shutil
import pandas as pd
```

```python
# Load the CSV file dynamically
source_name = 'Google' 
# source_name = 'ZWL' 
# source_name = 'Faith' 


# Define base path
base_path = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/Waste/img/'
QWen_Y = base_path + "Qwen_selected/" + source_name + '/'
QWen_N = base_path + "Qwen_Cpm_diff/" + source_name + '/'

csv_file_path = os.path.join(base_path, f'vl_{source_name}.csv')
df = pd.read_csv(csv_file_path)
df = df.fillna('')

df
```

```python
# Ensure both destination folders exist
os.makedirs(QWen_Y, exist_ok=True)
os.makedirs(QWen_N, exist_ok=True)

# Load the DataFrame (assuming it's already loaded as df)
# Define common image extensions
extensions = ['.jpg', '.JPG', '.jpeg', '.png', '.PNG']  # Add more if needed

for _, row in df.iterrows():
    img_name = row['img_name']
    img_dir = row['img_dir']
    domestic_value = row['Domestic']

    # Construct the full image path without extension
    img_path_without_ext = os.path.join(base_path, img_dir, img_name)
    
    # Try different extensions to find the actual image file
    img_path = None
    for ext in extensions:
        candidate_path = img_path_without_ext + ext
        if os.path.exists(candidate_path):
            img_path = candidate_path
            break
    
    # Determine the correct folder based on 'Domestic' value
    if domestic_value == 'Y':
        target_folder = QWen_Y
    elif domestic_value == 'N':
        target_folder = QWen_N
    else:
        continue  # Skip rows with missing or unexpected values

    # If image is found, copy it to the corresponding folder
    if img_path:
        shutil.copy(img_path, os.path.join(target_folder, os.path.basename(img_path)))
        print(f"Copied: {img_name}")
        # print(f"Copied: {img_path} -> {target_folder}")

    else:
        print(f"Image {img_name} not found in {os.path.join(base_path, img_dir)} with any supported extension.")
```

# Combine 3 to 1

```python
Base = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/Waste/img/'
df_F = pd.read_csv(Base + 'vl_Faith.csv')
df_G = pd.read_csv(Base + 'vl_Google.csv')
df_Z = pd.read_csv(Base + 'vl_ZWL.csv')

df_G
```

```python
# Check if all DataFrames have the same columns
def check_same_columns(*dfs):
    columns_set = [set(df.columns) for df in dfs]
    return all(columns_set[0] == cols for cols in columns_set)

# Usage
if check_same_columns(df_F, df_G, df_Z):
    print("All DataFrames have the same columns.")
else:
    print("DataFrames have different columns.")
```

```python
merged_df = pd.concat([df_F, df_G, df_Z], ignore_index=True)
merged_df
```

```python
merged_df.to_csv(Base + "Combined_SVI.csv", index=False)
```
