```python
import os
from pathlib import Path

def count_images_not_ending_with_angles(folder_path):
    """
    Count images in a folder that do NOT end with '_0', '_90', '_180', or '_270'
    
    Parameters:
    folder_path: str or Path - path to the folder containing images
    
    Returns:
    int - count of images that don't end with the specified suffixes
    """
    folder = Path(folder_path)
    
    # Common image extensions
    image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.tif', '.webp'}
    
    # Suffixes to exclude
    excluded_suffixes = {'_0', '_90', '_180', '_270'}
    
    count = 0
    matching_files = []
    
    # Iterate through all files in the folder
    for file_path in folder.iterdir():
        if file_path.is_file():
            # Check if it's an image file
            if file_path.suffix.lower() in image_extensions:
                # Get filename without extension
                stem = file_path.stem
                
                # Check if filename ends with any of the excluded suffixes
                ends_with_excluded = any(stem.endswith(suffix) for suffix in excluded_suffixes)
                
                if not ends_with_excluded:
                    count += 1
                    matching_files.append(file_path.name)
    
    return count, matching_files

# Example usage:
folder_path = "/Users/wenlanzhang/Downloads/PhD_UCL/Data/GoogleStreetView/Train/Train0315_695/images"
count, files = count_images_not_ending_with_angles(folder_path)
print(f"Number of images not ending with '_0', '_90', '_180', or '_270': {count}")
print(f"\nFirst 10 matching files:")
for f in files[:10]:
    print(f)
```
