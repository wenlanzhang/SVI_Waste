```python
import os
```

```python
# Set the directory where your images are stored
image_folder = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/UNDatathon/All_BG'

# List all files in the folder
image_files = [f for f in os.listdir(image_folder) if f.endswith(('.jpg', '.jpeg', '.png'))]

# Create an empty .txt file for each image
for image in image_files:
    # Get the image name without extension
    txt_filename = os.path.splitext(image)[0] + '.txt'
    # Create the empty .txt file in the same directory
    open(os.path.join(image_folder, txt_filename), 'w').close()

print(f"Created empty .txt files for {len(image_files)} images.")
```

```python
import os

# Set the directory where your .txt files are stored
txt_folder = '/Users/wenlanzhang/Downloads/PhD_UCL/Data/UNDatathon/Train_1027/labels'

# List all .txt files in the folder
txt_files = [f for f in os.listdir(txt_folder) if f.endswith('.txt')]

# Find non-empty .txt files that do not start with '0'
non_zero_start_files = []

for txt_file in txt_files:
    file_path = os.path.join(txt_folder, txt_file)
    
    # Read the file content
    with open(file_path, 'r') as file:
        lines = file.readlines()
        
        # Check if the file is not empty and does not start with '0'
        if lines and not lines[0].startswith('0'):
            non_zero_start_files.append(txt_file)

# Display the results
print(f"Found {len(non_zero_start_files)} non-empty .txt files that do not start with '0':")
for file in non_zero_start_files:
    print(file)
```
