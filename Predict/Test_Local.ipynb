```python
# Working

import os
import numpy as np

def check_and_correct_path(base_path, type_path, img_folder, modified_img_name):
    # List of possible extensions to try if the default doesn't exist
    possible_extensions = ['.JPG', '.jpg', '.png']
    for ext in possible_extensions:
        # Replace the extension in the modified image name
        name, _ = os.path.splitext(modified_img_name)
#         test_path = os.path.join(base_path, type_path, img_folder, name + ext)
        test_path = os.path.join(GSVIpath, img_folder, name + ext)
        if os.path.exists(test_path):
            return test_path  # Return the corrected path if file exists
    return None  # Return None if no valid path is found

for index, row in df_test.iterrows():

    try:
        img_folder = row['img_path']
        modified_img_name = row['img_name'] + '.JPG'
        source_path = os.path.join(Basepath, Typepath, img_folder, modified_img_name)

        # Check and correct the source path if necessary
        corrected_source_path = check_and_correct_path(Basepath, Typepath, img_folder, modified_img_name)
        if corrected_source_path is None:
            print(f"Error: Image file does not exist for index {index}. Tried multiple extensions.")
            raise FileNotFoundError("Valid image file not found after trying multiple extensions.")

        results = model(corrected_source_path)  # Run the YOLO model with the corrected path
        conf_list = results[0].boxes.cpu().conf.numpy().tolist()  # Get confidence scores as a list
        bbox_list = results[0].boxes.cpu().xyxy.numpy().tolist()  # Get bounding box information as a list

        df_test.at[index, 'conf'] = conf_list  # Store the confidence scores in 'conf' column
        df_test.at[index, 'bbox'] = bbox_list  # Store the bounding box information in 'bbox' column
    except Exception as e:
        print(f"Error processing index {index}: {e}")
        # Optionally, mark rows that failed processing for review
        df_test.at[index, 'conf'] = '-1'
        df_test.at[index, 'bbox'] = '-1'
df_test
```
